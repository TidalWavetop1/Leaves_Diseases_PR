```python
import os
import zipfile
from kaggle.api.kaggle_api_extended import KaggleApi

# 1. Khởi tạo và xác thực API
api = KaggleApi()
api.authenticate()

# 2. Định nghĩa tên dataset và thư mục lưu trữ
# Thay thế bằng tên dataset bạn muốn tải
dataset_name = "vipoooool/new-plant-diseases-dataset"
output_dir = "./data"

# Tạo thư mục lưu nếu chưa tồn tại
os.makedirs(output_dir, exist_ok=True)

print(f"**Bắt đầu tải dataset...**")
# 3. Tải dataset dưới dạng file .zip
api.dataset_download_files(dataset_name, path=output_dir, unzip=False)

# 4. Tìm file zip vừa tải và giải nén tự động bằng Python
for file in os.listdir(output_dir):
    if file.endswith(".zip"):
        zip_path = os.path.join(output_dir, file)
        print(f"Đang giải nén: {file}...")
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(output_dir)
        # Xóa file zip sau khi giải nén (tùy chọn để sạch ổ đĩa)
        os.remove(zip_path) 

print(f": Hoàn thành! Dữ liệu đã sẵn sàng trong thư mục: {output_dir}")
```
